```python
import pandas as pd
from pathlib import Path
from config import DATA_DIR

def load_apachejit(split: "train") -> pd.DataFrame:
    """Load an ApacheJIT split. split in {train, test_large, test_small, total}."""
    root = DATA_DIR / "raw/apachejit_dataset_replication/apachejit/dataset"
    paths = {
        "train": "apachejit_train.csv",
        "test_large": "apachejit_test_large.csv",
        "test_small": "apachejit_test_small.csv",
        "total": "apachejit_total.csv",
    }
    df = pd.read_csv(root / paths[split])
    assert df["buggy"].isin([True, False, 0, 1]).all()
    return df

df = load_apachejit("train")
print("train rows:", len(df),
      "buggy rate:", round(df["buggy"].mean(), 3))
```

```python
def engineer_features(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    # safe derivates
    df["total_lines"] = df["la"] + df["ld"]
    df["files_per_dir"] = df["nf"] / (df["nd"] + 1)
    df["change_density"] = (df["la"] + df["ld"]) / (df["nf"] + 1)
    df["author_file_exp_ratio"] = df["arexp"] / (df["aexp"] + 1)

    # encode the label
    df["buggy_int"] = df["buggy"].astype("int")

    feature_cols = [
        "la", "ld", "nf", "nd", "ns", "ent",
        "ndev", "age", "nuc", "aexp", "arexp", "asexp",
        "total_lines", "files_per_dir", "change_density", "author_file_exp_ratio",
    ]
    return df[feature_cols], df["buggy_int"]
```

```python
import joblib
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import average_precision_score

X_tr, y_tr = engineer_features(load_apachejit("train"))
X_te, y_te = engineer_features(load_apachejit("test_large"))

# Logistic regression needs scaling
scaler = StandardScaler().fit(X_tr)

lr = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42)
lr.fit(scaler.transform(X_tr), y_tr)

rf = RandomForestClassifier(n_estimators=300, class_weight="balanced", n_jobs=-1, random_state=42)
rf.fit(X_tr, y_tr)

# AUC-PR = average precision (your Week-1 headline number)
lr_auc_pr = average_precision_score(y_te, lr.predict_proba(scaler.transform(X_te))[:, 1])
rf_auc_pr = average_precision_score(y_te, rf.predict_proba(X_te)[:, 1])
print("LR  AUC-PR:", round(lr_auc_pr, 4))
print("RF  AUC-PR:", round(rf_auc_pr, 4))
```

```python
def recall_at_k(y_true, scores, k=0.20):
    cutoff = int(len(y_true) * k)
    order = scores.argsort()[::-1]
    top = order[:cutoff]
    return y_true.iloc[top].sum() / y_true.sum()

print("LR  recall@20%:", round(recall_at_k(y_te, lr.predict_proba(scaler.transform(X_te))[:, 1]), 3))
print("RF  recall@20%:", round(recall_at_k(y_te, rf.predict_proba(X_te)[:, 1]), 3))
```

```python
from config import MODELS_DIR
MODELS_DIR.mkdir(parents=True, exist_ok=True)
joblib.dump((scaler, lr), MODELS_DIR / "baseline_lr.joblib")
joblib.dump(rf, MODELS_DIR / "baseline_rf.joblib")
# log the Week-1 numbers
import json
(MODELS_DIR / "baseline_metrics.json").write_text(
    json.dumps({"lr_auc_pr": lr_auc_pr, "rf_auc_pr": rf_auc_pr}, indent=2))
```
